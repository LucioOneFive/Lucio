# Fully loaded R&D cost walk-down (waterfall plot)
This notebook recreates the waterfall plot shown in the prompt.

**Notes**
- Data is hard-coded.
- Intermediate bars are shown as floating (step) bars.


```
import matplotlib.pyplot as plt

# -----------------------------
# Hard-coded data (EUR)
# -----------------------------
traditional_total = 451_428
material_production_cost = 43_151
trial_cost = 92_027
rd_labor_costs = 153_333

# Final total as shown in the figure
reverse_engineering_total = 162_916

labels = [
    "Traditional total",
    "Material production cost",
    "Trial cost",
    "R&D labor costs",
    "Reverse engineering total",
]

# Cost components to walk down from the traditional total
components = [material_production_cost, trial_cost, rd_labor_costs]

# Colors chosen to match the example image
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd"]
```

```
def plot_waterfall(traditional_total, components, final_total, labels, colors):
    """Create a basic waterfall plot with floating intermediate bars."""
    # Cumulative walk-down (start + intermediate endpoints)
    cum = [traditional_total]
    for v in components:
        cum.append(cum[-1] - v)

    # Build bar bottoms/heights
    bottoms = [0]
    heights = [traditional_total]

    # Floating bars for each component
    for i, v in enumerate(components, start=1):
        # bar spans from new cumulative (after subtract) up to previous cumulative
        bottom = cum[i]
        height = v
        bottoms.append(bottom)
        heights.append(height)

    # Final total bar
    bottoms.append(0)
    heights.append(final_total)

    x = list(range(len(labels)))

    fig, ax = plt.subplots(figsize=(12, 5))
    ax.bar(x, heights, bottom=bottoms, color=colors, width=0.8)

    ax.set_title("Fully loaded R&D cost walk-down (actual data)")
    ax.set_ylabel("Cost (€)")
    ax.set_xticks(x)
    ax.set_xticklabels(labels, rotation=20)

    # Annotate values in the middle of each bar
    values_for_labels = [traditional_total] + components + [final_total]
    for xi, btm, h, v in zip(x, bottoms, heights, values_for_labels):
        ax.text(xi, btm + h / 2, f"{v} €", ha="center", va="center", fontsize=10)

    plt.tight_layout()
    return fig, ax

fig, ax = plot_waterfall(
    traditional_total=traditional_total,
    components=components,
    final_total=reverse_engineering_total,
    labels=labels,
    colors=colors,
)
plt.show()
```
